```python
import speech_recognition as sr
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

# Diccionario de comandos de color
commands = {
    "rojo": (1, 0, 0),
    "verde": (0, 1, 0),
    "azul": (0, 0, 1),
    "amarillo": (1, 1, 0),
    "blanco": (1, 1, 1),
    "negro": (0, 0, 0)
}

# Función para mostrar un gráfico con el color
def show_color(color_rgb, label):
    plt.figure(figsize=(3, 3))
    plt.gca().add_patch(Rectangle((0, 0), 1, 1, color=color_rgb))
    plt.xticks([])
    plt.yticks([])
    plt.title(f"Color: {label}", fontsize=16)
    plt.show()

# Inicializar el reconocedor y micrófono
recognizer = sr.Recognizer()
mic = sr.Microphone()

# Bucle de reconocimiento
while True:
    try:
        with mic as source:
            print("\nHabla un color (rojo, verde, azul...) o 'salir' para terminar:")
            recognizer.adjust_for_ambient_noise(source)
            audio = recognizer.listen(source)

        # Reconocimiento con Sphinx
        recognized = recognizer.recognize_sphinx(audio, language="es-ES")
        recognized = recognized.lower()
        print(f"Reconocido: {recognized}")

        if "salir" in recognized:
            print("Terminando...")
            break

        for palabra in commands:
            if palabra in recognized:
                color = commands[palabra]
                show_color(color, palabra)
                break
        else:
            print("Comando no reconocido.")

    except sr.UnknownValueError:
        print("No se entendió el audio.")
    except sr.RequestError as e:
        print(f"Error con el reconocimiento: {e}")
```
