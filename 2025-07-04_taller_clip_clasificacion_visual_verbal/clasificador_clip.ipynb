```python
# Importar librerías
import clip
import torch
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np
import os
import requests
import imageio
from google.colab import drive
from getpass import getpass

# Habilitar visualización en Colab
%matplotlib inline

# Montar Google Drive (opcional, para guardar imágenes y resultados de forma persistente)
mount_drive = input("¿Quieres montar Google Drive? (s/n): ")
if mount_drive.lower() == 's':
    drive.mount('/content/drive')
    base_dir = '/content/drive/MyDrive/2025-06-28_taller_clip_clasificacion_visual_verbal'
else:
    base_dir = '/content/2025-06-28_taller_clip_clasificacion_visual_verbal'

# Configurar directorios
image_dir = os.path.join(base_dir, 'images')
output_dir = os.path.join(base_dir, 'resultados')

# Crear directorios si no existen
os.makedirs(image_dir, exist_ok=True)
os.makedirs(output_dir, exist_ok=True)

# Configurar dispositivo (GPU si está disponible, de lo contrario CPU)
device = "cuda" if torch.cuda.is_available() else "cpu"

# Cargar modelo CLIP y preprocesador
model, preprocess = clip.load("ViT-B/32", device=device)

def download_images(query, api_key, count=3, output_dir=image_dir):
    """Descargar imágenes de Unsplash usando la API."""
    url = f"https://api.unsplash.com/search/photos?query={query}&per_page={count}&client_id={api_key}"
    try:
        response = requests.get(url).json()
        image_paths = []
        for i, photo in enumerate(response['results']):
            img_url = photo['urls']['regular']
            img_data = requests.get(img_url).content
            img_path = os.path.join(output_dir, f"{query}_{i}.jpg")
            with open(img_path, "wb") as f:
                f.write(img_data)
            image_paths.append(img_path)
        return image_paths
    except Exception as e:
        print(f"Error al descargar imágenes para '{query}': {e}")
        return []

def load_images(image_dir):
    """Cargar todas las imágenes del directorio especificado."""
    image_paths = [os.path.join(image_dir, f) for f in os.listdir(image_dir) if f.endswith(('.jpg', '.png', '.jpeg'))]
    images = [preprocess(Image.open(path)).unsqueeze(0).to(device) for path in image_paths]
    return image_paths, torch.cat(images)

def classify_images(image_paths, images, text_descriptions):
    """Clasificar imágenes usando CLIP con descripciones de texto."""
    # Tokenizar descripciones de texto
    text = clip.tokenize(text_descriptions).to(device)

    # Obtener embeddings de imágenes y texto
    with torch.no_grad():
        image_features = model.encode_image(images)
        text_features = model.encode_text(text)
        logits_per_image, _ = model(images, text)
        probs = logits_per_image.softmax(dim=-1).cpu().numpy()

    return probs

def visualize_results(image_paths, probs, text_descriptions, output_dir):
    """Visualizar resultados de clasificación para cada imagen."""
    result_images = []
    for i, (image_path, prob) in enumerate(zip(image_paths, probs)):
        # Cargar y mostrar imagen
        img = Image.open(image_path)
        plt.figure(figsize=(10, 6))
        plt.subplot(1, 2, 1)
        plt.imshow(img)
        plt.axis('off')
        plt.title(f"Imagen: {os.path.basename(image_path)}")

        # Graficar probabilidades
        plt.subplot(1, 2, 2)
        plt.barh(text_descriptions, prob)
        plt.xlabel('Probabilidad')
        plt.title('Probabilidades de Clasificación')
        plt.tight_layout()

        # Guardar gráfica
        output_path = os.path.join(output_dir, f"result_{os.path.basename(image_path)}.png")
        plt.savefig(output_path)
        result_images.append(output_path)
        plt.show()  # Mostrar en Colab

        # Imprimir clase más probable
        max_prob_idx = np.argmax(prob)
        print(f"Imagen: {os.path.basename(image_path)}")
        print(f"Clase más probable: {text_descriptions[max_prob_idx]} ({prob[max_prob_idx]:.2%})")
        print("Todas las probabilidades:", {desc: f"{p:.2%}" for desc, p in zip(text_descriptions, prob)})
        print("-" * 50)

    # Generar GIF con los resultados
    gif_path = os.path.join(output_dir, "results.gif")
    with imageio.get_writer(gif_path, mode='I', duration=1) as writer:
        for img_path in result_images:
            image = imageio.imread(img_path)
            writer.append_data(image)
    print(f"GIF guardado en: {gif_path}")

def main():
    # Solicitar clave de API de Unsplash de forma segura
    unsplash_api_key = getpass("Ingresa tu clave de API de Unsplash: ")

    # Descargar imágenes usando la API de Unsplash
    queries = ["cat", "dog", "car", "tree"]
    all_image_paths = []
    for query in queries:
        image_paths = download_images(query, unsplash_api_key, count=2)
        all_image_paths.extend(image_paths)

    # Cargar imágenes
    image_paths, images = load_images(image_dir)

    # Descripciones de texto para clasificación
    text_descriptions = [
        "un gato",
        "un perro",
        "un caballo",
        "un coche",
        "un árbol",
        "un gato borroso",
        "un coche deportivo rápido",
        "algo feliz",
        "un objeto peligroso"
    ]

    # Clasificar imágenes
    probs = classify_images(image_paths, images, text_descriptions)

    # Visualizar y guardar resultados
    visualize_results(image_paths, probs, text_descriptions, output_dir)

if __name__ == "__main__":
    main()
```

```python
import os
import zipfile
from google.colab import files

base_dir = '/content/2025-06-28_taller_clip_clasificacion_visual_verbal'
images_dir = os.path.join(base_dir, 'images')
resultados_dir = os.path.join(base_dir, 'resultados')
output_zip = os.path.join(base_dir, 'taller_output.zip')

# Comprimir las carpetas images/ y resultados/
if os.path.exists(images_dir) and os.path.exists(resultados_dir):
    with zipfile.ZipFile(output_zip, 'w', zipfile.ZIP_DEFLATED) as zipf:
        # Comprimir carpeta images/
        for root, _, files in os.walk(images_dir):
            for file in files:
                file_path = os.path.join(root, file)
                zipf.write(file_path, os.path.relpath(file_path, base_dir))
        # Comprimir carpeta resultados/
        for root, _, files in os.walk(resultados_dir):
            for file in files:
                file_path = os.path.join(root, file)
                zipf.write(file_path, os.path.relpath(file_path, base_dir))
    print(f"Archivo ZIP creado: {output_zip}")

else:
    print(f"Una o ambas carpetas no existen: {images_dir}, {resultados_dir}")
```
